# Ablation 3 Counterfactual: Multi-Scale Only (No Augmentations)

**Research Question**: Open: is multi-scale training effective without augmentations.

**Hypothesis**: Disabling augmentations isolates the multi-scale effect and tests whether the RandAugment and multi-scale conflict caused the 95% failure.

**Implementation**:
- Multi-scale: [416, 512, 640] pixels (SAME as original A3)
- All augmentations: explicitly disabled (workaround for the auto_augment=None bug)

**Expected Outcome**:
- If Dice improves: Multi-scale works, augmentation conflict was the issue
- If Dice fails: Multi-scale itself is problematic

**Fix**: GitHub Issue #19815 shows `auto_augment=None` does not work. All 15 augmentation parameters are set to 0.0.

---

```python
# Multi-seed training with Multi-Scale ONLY (P100-Optimized)
RANDOM_SEEDS = [42, 123, 456]

# P100-Optimized Hyperparameters (SAME as baseline for fair comparison)
BATCH_SIZE = 32
LEARNING_RATE = 0.015
WORKERS = 8
AMP = False

# Dataset-Aligned Parameters
IMAGE_SIZE = 512
EPOCHS = 100
PATIENCE = 10

# Storage for all training results
all_training_results = {}

print("=" * 80)
print("ABLATION 3 COUNTERFACTUAL: MULTI-SCALE ONLY (NO AUGMENTATIONS)")
print("=" * 80)
print(f"\nTraining configuration:")
print(f"  Seeds: {RANDOM_SEEDS}")
print(f"  Batch size: {BATCH_SIZE}")
print(f"  Learning rate: {LEARNING_RATE}")
print(f"  Multi-scale: [416, 512, 640] pixels (random per batch)")
print(f"  ALL AUGMENTATIONS: EXPLICITLY DISABLED (15 parameters set to 0.0)")
print(f"\nDataset config: {config_path}")
print(f"\nEstimated time: ~1.5-2 hours total (~30-40 min per seed on P100)\n")

for seed_idx, seed in enumerate(RANDOM_SEEDS, 1):
    print("\n" + "=" * 80)
    print(f"TRAINING WITH SEED {seed} ({seed_idx}/{len(RANDOM_SEEDS)})")
    print("=" * 80)

    # Initialize model from pretrained COCO weights
    model = YOLO("yolov8n-seg.pt")

    # Train model
    training_start = time.time()

    results = model.train(
        data=str(config_path),
        # P100-optimized parameters
        batch=BATCH_SIZE,
        lr0=LEARNING_RATE,
        workers=WORKERS,
        amp=AMP,
        # Dataset-aligned
        epochs=EPOCHS,
        patience=PATIENCE,
        # Multi-Scale Training (INTERVENTION - SAME as original A3)
        imgsz=[416, 512, 640],  # Multiple scales
        # CRITICAL DIFFERENCE: EXPLICITLY DISABLE ALL AUGMENTATIONS
        # Workaround for auto_augment=None bug (GitHub Issue #19815)
        hsv_h=0.0,          # Hue augmentation
        hsv_s=0.0,          # Saturation augmentation
        hsv_v=0.0,          # Value augmentation
        degrees=0.0,        # Rotation
        translate=0.0,      # Translation
        scale=0.0,          # Scale
        shear=0.0,          # Shear
        perspective=0.0,    # Perspective
        flipud=0.0,         # Vertical flip
        fliplr=0.0,         # Horizontal flip
        mosaic=0.0,         # Mosaic augmentation
        mixup=0.0,          # Mixup augmentation
        copy_paste=0.0,     # Copy-paste augmentation
        erasing=0.0,        # Random erasing
        auto_augment=None,  # Disable AutoAugment (plus explicit disabling above)
        # Reproducibility
        seed=seed,
        save=True,
        project=str(OUTPUT_PATH / "yolo_training"),
        name=f"seed_{seed}",
        exist_ok=True,
        verbose=False,
    )

    training_duration = time.time() - training_start
    print(f"\nOK Training completed in {training_duration/60:.1f} minutes")

    # EVALUATION
    print(f"\nEvaluating model on test set ({len(test_pairs)} images)...")

    seed_results = []
    for idx, (img_path_str, mask_path_str) in enumerate(test_pairs):
        # Extract image ID from path string
        image_id = Path(img_path_str).stem

        # Load image
        img = cv2.imread(img_path_str)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        # Load ground truth mask
        gt_mask = cv2.imread(mask_path_str, cv2.IMREAD_GRAYSCALE)
        _, gt_mask = cv2.threshold(gt_mask, 127, 1, cv2.THRESH_BINARY)

        # Run inference
        yolo_results = model.predict(img_rgb, verbose=False)

        # Convert YOLO result to metrics input format
        conversion_result = yolo_result_to_metrics_input(yolo_results[0], mask_path_str, image_id)

        if conversion_result["success"]:
            pred_mask = conversion_result["data"]["pred_mask"]
            gt_mask_from_result = conversion_result["data"]["gt_mask"]

            # Calculate segmentation metrics
            metrics = calculate_segmentation_metrics(pred_mask, gt_mask, image_id)

            # Calculate boundary F1
            boundary_f1 = calculate_boundary_f1(pred_mask, gt_mask, tolerance=2)
            metrics["boundary_f1"] = boundary_f1

            seed_results.append(metrics)
        else:
            print(f"  WARNING: Failed to process {image_id}: {conversion_result['error']}")

        # Progress update
        if (idx + 1) % 100 == 0:
            print(f"  Processed {idx + 1}/{len(test_pairs)} images...")

    # Store results for this seed
    all_training_results[f"seed_{seed}"] = seed_results

    print(f"\nOK Seed {seed} complete: {len(seed_results)} images evaluated")

print("\n" + "=" * 80)
print("ALL TRAINING/EVALUATION RUNS COMPLETE")
print("=" * 80)
print(f"Total seeds: {len(all_training_results)}")
print(f"Images per seed: {len(next(iter(all_training_results.values())))}")
```

## Aggregate Results and Statistical Validation

**Aggregation**: Pool per-image predictions across the 3 seeds.

**Statistical Analysis**:
- Bootstrap CIs (10,000 iterations) for each metric
- Cross-seed variance quantification
- Comparison to baseline (80.0% Dice)

**Clinical Significance**: Low cross-seed variance (CV < 2%) validates reproducibility.

```python
# Aggregate results across all seeds
print("=" * 80)
print("AGGREGATING RESULTS")
print("=" * 80)

# Pool all results from all seeds
all_results = []
for seed_key, seed_results in all_training_results.items():
    all_results.extend(seed_results)

print(f"\nTotal predictions pooled: {len(all_results)}")
print(f"Expected: {len(test_pairs) * len(RANDOM_SEEDS)}")

# Calculate aggregate statistics with bootstrap CIs
dice_scores = [r["dice"] for r in all_results]
iou_scores = [r["iou"] for r in all_results]
precision_scores = [r["precision"] for r in all_results]
recall_scores = [r["recall"] for r in all_results]
boundary_f1_scores = [r["boundary_f1"] for r in all_results]

print(f"\nCalculating bootstrap confidence intervals...\n")

summary_stats = {}

for metric_name, metric_values in [
    ("dice", dice_scores),
    ("iou", iou_scores),
    ("precision", precision_scores),
    ("recall", recall_scores),
    ("boundary_f1", boundary_f1_scores),
]:
    ci_result = calculate_bootstrap_ci(
        metric_values, confidence_level=0.95, n_bootstrap=10000, random_seed=42
    )
    summary_stats[metric_name] = ci_result

    # Compare to baseline
    baseline_mean = 0.8000
    improvement = ci_result["mean"] - baseline_mean

    print(f"{metric_name.upper()}:")
    print(f"  Counterfactual: {ci_result['mean']:.4f} ± {ci_result['std']:.4f}")
    print(f"  95% CI: [{ci_result['ci_lower']:.4f}, {ci_result['ci_upper']:.4f}]")
    print(f"  Baseline:       {baseline_mean:.4f}")
    print(f"  Improvement:    +{improvement:.4f} ({100*improvement/baseline_mean:.1f}%)")
    print()

print("=" * 80)
print("COUNTERFACTUAL ANALYSIS COMPLETE")
print("=" * 80)

# Calculate cross-seed variance
print(f"\nCross-Seed Performance Variance:\n")

for seed_key, seed_results in all_training_results.items():
    seed_dice_mean = np.mean([r["dice"] for r in seed_results])
    print(f"  {seed_key}: Dice = {seed_dice_mean:.4f}")

# Calculate coefficient of variation
seed_means = [np.mean([r["dice"] for r in results]) for results in all_training_results.values()]
cv = 100 * np.std(seed_means) / np.mean(seed_means)

print(f"\n  Coefficient of Variation: {cv:.2f}%")
if cv < 2.0:
    print(f"  OK Low variance - results are robust across random seeds")
else:
    print(f"  WARNING High variance - consider additional seeds for stability")
```
